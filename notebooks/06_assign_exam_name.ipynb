```python
files = [i for i in EXAMS_PATH.rglob("*.pdf")]
```

```python
for file in files:
    filename = file.name
    path_parts = file.parts
    year = int(filename[:4])

    # orden de reemplazo es importante!!!
    if year <= 2020:
        new_filename = (filename
                        .replace("extra ", "septiembre ")
                        .replace("extraordinaria ", "septiembre ")
                        .replace("ordinaria ", "junio ")
                        .replace("ord ", "junio ")
                    
        )
    else:
        new_filename = (filename
                        .replace("extra ", "julio ")
                        .replace("extraordinaria ", "julio ")
                        .replace("ordinaria ", "junio ")
                        .replace("ord ", "junio ")     
        )

    # remove duplicates
    filename_words = new_filename.split(" ")
    deduplicated_filename_words = []
    for w in filename_words:
        if w not in deduplicated_filename_words:
            deduplicated_filename_words.append(w)
            
    new_filename = "_".join(deduplicated_filename_words)

    # replace triple hyphen
    new_filename = new_filename.replace("_-_", "_").replace("_.pdf", ".pdf")

    new_file = "/".join(path_parts[:-1])
    new_file = f"{new_file}/{new_filename}"

    shutil.move(file, new_file)     
```
